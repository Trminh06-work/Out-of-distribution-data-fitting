# Import libraries

```python
import os
import pandas as pd
import numpy as np
import random
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split

from random_split import random_split
from modified_mfs_split import main # Import metafeature-based split method
from geometric_split import GeometricSplit
from marginal_distribution_split import MarginalDistributionSplit
```

```python
path_prefix = "../data/splitted/"
```

```python
dataset_list = [
    "218_house_8L", "house", "california", "elevators", "CASP", "gas_turbine_co_and_nox_emission", 
    "CBM_1", "CBM_2", "delays_zurich_transport", "diamonds", "nyc_taxi_green_dec_2016", "house_sales", "medical_charges", 
    "MiamiHousing2016", "sulfur_1", "sulfur_2", "books", "players_22", "taxi", "bike", 
    "power_consumption", "3droad", "keggdirected", "kin40k", "protein", "tamielectric"
]
```

```python
TRAIN_SIZE = 0.7
TEST_SIZE = 0.3
SEEDS = [0, 42, 19, 2, 23, 11, 37, 29, 57, 5] # 10 random SEEDs to construct CI
```

# Marginal distribution-based Splitting

```python
for file_name in dataset_list:
    print(f"Start processing {file_name}.parquet...")
    
    data_path = f"../data/raw/{file_name}/{file_name}.parquet"
    df = pd.read_parquet(data_path)
    margin_splitter = MarginalDistributionSplit(file_name, df, TEST_SIZE, seeds = SEEDS)
    margin_splitter.covariate_shift()
    
    print(f"Successfully splitted!!!")
    print("==" * 25)

print("Complete Marginal distribution-based Partition")
```

# Geometric-based Splitting

```python
for file_name in dataset_list:
    print(f"Start processing {file_name}.parquet...")
    
    data_path = f"../data/raw/{file_name}/{file_name}.parquet"
    df = pd.read_parquet(data_path)
    geometry_splitter = GeometricSplit(file_name, df, TEST_SIZE, SEEDS)

    # Single Hyperball
    geometry_splitter.single_hyperball()
    
    # Multiple Hyperballs
    geometry_splitter.multiple_hyperballs(num_balls = 5)
    
    # Single Slab
    geometry_splitter.single_slab()
    
    # Semi-infinite Slab
    geometry_splitter.semi_infinite_slab()
    
    # KMeans-based Hyperballs
    geometry_splitter.kmeans_hyperballs(n_clusters = 5)

    print(f"Successfully splitted!!!")
    print("==" * 25)

print("Complete Geometric-based Partition")
```

# Random Train/Test Split

```python
for file_name in dataset_list:
    print(f"Start processing {file_name}.parquet...")

    random_split(file_name, SEEDS, TEST_SIZE)
    
    print(f"Successfully splitted!!!")
    print("==" * 25)

print("Complete Random Train/Test Partition")
```

# Metafeature-based Splitting

This approach is adopted from Paper ["Evaluating robustness of tabular models under meta-features based shifts"](https://openreview.net/pdf?id=BS68QcFppq), and is a replication of the code found in the following GitHub repository:
[OOD_Tab_Evaluation](https://github.com/ITMO-NSS-team/OOD_Tab_Evaluation.git)

```python
for file_name in dataset_list:
    path = f"../data/raw/{file_name}/{file_name}.parquet"

    folder_path = f"../data/splitted/{file_name}/Mfs_based_Split"

    if os.listdir(folder_path):     # list is non-empty → True
        print(f"Skip {file_name}.parquet due to splitted data exist")
        continue
    
    df = pd.read_parquet(path)
        
    # aim = mutual information (mut_inf)
    print(f"Start processing {file_name}.parquet...")
    main(file_name = file_name,
          aim = "mut_inf",
          seeds = SEEDS,
          keep_size = False       # Randomly select 800,000 samples for extensive datasets, i.e. >1M samples
        )
    print(f"Successfully splitted!!!")
    print("==" * 25)

print("Complete Metafeature (mutual information)-based Partition")
```
